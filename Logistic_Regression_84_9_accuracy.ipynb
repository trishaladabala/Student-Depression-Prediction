```python
df = pd.read_csv('/kaggle/input/student-depression-dataset/student_depression_dataset.csv')
df.drop(columns=['id'], inplace=True, errors='ignore')
cate_cols = df.select_dtypes(include=['object']).columns.tolist()
```

```python
from sklearn.preprocessing import LabelEncoder
cate_cols=df.select_dtypes(include=['object']).columns.tolist()
cate_cols = ['Gender', 'City', 'Profession', 'Degree', 'Dietary Habits',
                    'Sleep Duration', 'Have you ever had suicidal thoughts ?',
                    'Family History of Mental Illness']

le=LabelEncoder()
for col in cate_cols:
    df[col] = df[col].astype(str).str.strip("'")
    df[col] = le.fit_transform(df[col])
```

```python
df.fillna(df.median(numeric_only=True), inplace=True)
X = df.drop(columns=['Depression'])
y = df['Depression']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
from sklearn.impute import SimpleImputer
X_train.replace('?', np.nan, inplace=True)
X_test.replace('?', np.nan, inplace=True)
imputer = SimpleImputer(strategy='median')
X_train = imputer.fit_transform(X_train)
X_test = imputer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0,max_iter=1000)
classifier.fit(X_train, y_train)
y_pred = classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_pred, y_test))
```
